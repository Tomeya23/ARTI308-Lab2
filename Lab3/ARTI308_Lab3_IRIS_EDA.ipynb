```
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("iris.csv")

df.head()
```

## 1. Basic Data Inspection

```
print("Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nMissing Values:\n", df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nDescriptive Statistics:\n", df.describe())
```

## 2. Univariate Analysis

```
plt.figure()
plt.hist(df['sepal_length'])
plt.title("Sepal Length Distribution")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

```
plt.figure()
plt.boxplot(df['petal_length'])
plt.title("Petal Length Boxplot")
plt.show()
```

## 3. Bivariate Analysis

```
plt.figure()
plt.scatter(df['sepal_length'], df['petal_length'])
plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```
correlation = df.corr(numeric_only=True)
print("Correlation Matrix:\n", correlation)
```

## 4. Conclusion

- The dataset has no missing values.
- Features are normally distributed with slight variation.
- Strong positive correlation exists between petal length and petal width.
- Iris dataset is clean and suitable for classification tasks.
